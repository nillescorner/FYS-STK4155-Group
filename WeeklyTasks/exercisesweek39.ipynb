## Exercise 3: the conclusion

The conclusion should

* state your main findings and interpretations,
* try as far as possible to present perspectives for future work,
* try to discuss the pros and cons of the methods and possible improvements.

Write the conclusion of your Project 1 report in three paragraphs.

**3a)** The main findings, explained. The numbers of the abstract, now with
the reason behind them: why is the test error minimised where it is, why does
Ridge extend the usable range of degrees, why does Lasso set the coefficients
it sets to zero, why does gradient descent need standardised features. No
sentence should repeat the abstract verbatim.

**3b)** The critical paragraph. What the methods could and could not do, what
you would do differently, and where your results are least certain (for
instance a learning-rate grid too coarse to rank two optimisers, a fixed
iteration count instead of a tolerance, a bias estimate that uses the known
function). Write this paragraph first; it is the hardest and the most
valuable, and a limitation you state yourself raises the score under
"critical comments", while one the grader finds lowers it.

**3c)** Perspectives. The one or two things you would do next and why,
concretely ("the Lasso path with warm starts along $\lambda$", "a
